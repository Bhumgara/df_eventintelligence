# Skiddle API EDA

```python
from apis import skiddle as sk
from pprint import pprint

import pandas as pd
from math import ceil
```

```python
endpoint = 'events/search/'
filters = {
    'country': 'GB',
    'eventcode': 'FEST'
}

results = sk.fetch_all_events(filters,verbose=True)
len(results)

# data = sk.query(endpoint, filters=filters, verbose=True)
# results = data['results']
```

```python
keys = ['id','listingid','eventname','startdate','enddate']
df = pd.DataFrame([{key: r.get(key, None) for key in keys} for r in results])
df['startdate'] = pd.to_datetime(df['startdate'])
df['enddate'] = pd.to_datetime(df['enddate'])

print(f"Total rows: {len(df)}")
df.head()
```

```python
endpoint = 'events/search/'
filters = {
    'country': 'GB',
    'eventcode': 'FEST'
}
raw_events = sk.query_all(endpoint, filters=filters, verbose=True)

events = sk.validate_events(raw_events)
df_events = sk.events_to_df(events)
df_events.head()
```
